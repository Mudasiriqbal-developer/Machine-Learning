# Multipe Linear Regression


```python
import pandas as pd
```

```python
dataset=pd.read_csv('D:\Machine Learning\Practice in ML\iris_multiple_linear_reg_dataset.csv')
```

```python
dataset
```

```python
dataset['Species'].unique()
```

```python
dataset['Species']=dataset['Species'].replace({'Iris-setosa':'1', 'Iris-versicolor': '2', 'Iris-virginica':'3'})
```

```python
dataset
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(
  dataset[
     ['SepalLengthCm',	'SepalWidthCm',	'PetalLengthCm',	'PetalWidthCm']
    ],dataset[
        'Species'
      ], 
      test_size = 0.2
)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
```

```python
lr.fit(x_train, y_train)
```

```python
lr.predict(x_test)
```

```python
x_test   #{'Iris-setosa':'1', 'Iris-versicolor': '2', 'Iris-virginica':'3'}
```

```python
lr.score(x_test, y_test)
```
